```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

# reading image

```python
image=cv2.imread(r"C:\Users\snaha\Desktop\image_processing\Lenna_(test_image).png",0)
if image is None:
    print("Image not found!")
else:
    print("Image loaded successfully.")
```

# Display Image Using OpenCV

```python
cv2.imshow("Grayscale Image", image)
plt.imshow(image, cmap="gray")
```

```python
plt.figure(figsize=(5,5))
plt.imshow(image, cmap="gray")
plt.title("Grayscale Lena Image")
plt.axis("off")
plt.show()
```

# Find Image Properties

```python
height, width = image.shape

number_of_pixels = image.size

datatype = image.dtype

minimum = image.min()

maximum = image.max()

mean = image.mean()

print("Image Height :", height)
print("Image Width  :", width)
print("Number of Pixels :", number_of_pixels)
print("Datatype :", datatype)
print("Minimum Intensity :", minimum)
print("Maximum Intensity :", maximum)
print("Mean Intensity :", mean)
```

```python
_, binary = cv2.threshold(image,127, 255, cv2.THRESH_BINARY)
```

```python
plt.imshow(binary,cmap="grey")
plt.title("Binary Image")
plt.axis("off")
plt.show()
```
