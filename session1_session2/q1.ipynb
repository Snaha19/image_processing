```python
import cv2
import matplotlib.pyplot as plt

# Read grayscale image
image = cv2.imread("lena.png", 0)

# a. Increase brightness
bright = cv2.convertScaleAbs(image, alpha=1, beta=50)

# b. Decrease brightness
dark = cv2.convertScaleAbs(image, alpha=1, beta=-50)

# c. Increase contrast
high_contrast = cv2.convertScaleAbs(image, alpha=1.5, beta=0)

# d. Decrease contrast
low_contrast = cv2.convertScaleAbs(image, alpha=0.5, beta=0)

# e. Negative image
negative = 255 - image

# f. Binary thresholding
_, binary = cv2.threshold(image, 127, 255, cv2.THRESH_BINARY)

# Display all images
titles = [
    "Original",
    "Increase Brightness",
    "Decrease Brightness",
    "Increase Contrast",
    "Decrease Contrast",
    "Negative",
    "Binary Threshold"
]

images = [
    image,
    bright,
    dark,
    high_contrast,
    low_contrast,
    negative,
    binary
]

plt.figure(figsize=(12, 8))

for i in range(len(images)):
    plt.subplot(3, 3, i + 1)
    plt.imshow(images[i], cmap="gray")
    plt.title(titles[i])
    plt.axis("off")

plt.tight_layout()
plt.show()
```
